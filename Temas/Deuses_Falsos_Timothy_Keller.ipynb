## Deuses do lucro
O argumento tradicional que o único modo de promover o cbem comum de um negócio é criar empregos e gerar novos produtos.

Em algumas situações mercado tira dinheiro de quem possui falta de integridade ou é desonesto. Em outras, a honestidade pode ser financeiramente prejudicial.

O **lucro** quando considerado como valor absoluto, portanto, pode ser considerado um *falso deus*, resultando em um colapso moral e social.
