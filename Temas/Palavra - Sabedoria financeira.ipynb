## Registro do planejamento

Planos são etéreos, está na sua mente, mas podem ser esquecidos, Quando Deus concorda com seus planejamento, você deve deixá-lo de forma visível:

> ² Então o Senhor me respondeu, e disse: Escreve a visão e torna-a bem legível sobre tábuas, para que a possa ler quem passa correndo. (Habacuque 2:2 ACF)

> ¹¹ Porque a pedra clamará da parede, e a trave lhe responderá do madeiramento. (Habacuque 2:11 ACF)

Seja claro e estabeleça com metas expressas de tempo, custos e descrições.

Metas devem ser viáveis, pois aquelas difíceis de manter logo se enfraquecerão na primeira falha.

Se possível, subdivida com submetas, pois reforçam as chances de seguir em direção à meta principal.

Vê-las com frequência ajudará a se lembrar por que você está sacrificando seu conforto e ti impedirá de recair em más decisões por impulso.

Em Lucas, Jesus comparou realizar um planejamento com é como segui-lo. Todo ato incial se desdobra em várias consequências e compromissos profundos:

> ²⁷ E qualquer que não levar a sua cruz, e não vier após mim, não pode ser meu discípulo. ²⁸ Pois qual de vós, querendo edificar uma torre, não se assenta primeiro a fazer as contas dos gastos, para ver se tem com que a acabar?
²⁹ Para que não aconteça que, depois de haver posto os alicerces, e não a podendo acabar, todos os que a virem comecem a escarnecer dele,
³⁰ Dizendo: Este homem começou a edificar e não pôde acabar.
³¹ Ou qual é o rei que, indo à guerra a pelejar contra outro rei, não se assenta primeiro a tomar conselho sobre se com dez mil pode sair ao encontro do que vem contra ele com vinte mil?
³² De outra maneira, estando o outro ainda longe, manda embaixadores, e pede condições de paz. (Lucas 14:27-32 ACF)

Fazer as contas dos gastos de uma construção e planejar uma guerra equivalem a ter plena conciência que depois de iniciado o projeto, muitos prejuízos. Esses prejuízos são literalmente referente a materiais na parábola. Ao interpretar o sentido da parábola, é dito que se não houver uma decisão madura para segui-lo, poderá vir a frustrar-se no futuro.

## Problemas do excesso de planejamento

Um lenhador não recebe seu salário apenas por afiar seu machado. Apesar de este ser necessário para cortar a árvore e que um bom fio facilitará o trabalho duro, mas somente o planejamento não assegura a execução.

>⁴ Quem observa o vento, nunca semeará, e o que olha para as nuvens nunca segará.
⁵ Assim como tu não sabes qual o caminho do vento, nem como se formam os ossos no ventre da mulher grávida, assim também não sabes as obras de Deus, que faz todas as coisas. (Eclesiastes 11:4,5 ACF)

A paralisia do planejamento, a espera longa de um momento perfeito para iniciar a execução, ou ainda a baixa auto-estima são aspectos que influenciam na morte de um bom plano neste estágio.

O tempo ideal pode não existir. Deus nos capacita e nos orienta sobre nossos planos.

Nem sempre uma situação difícil ocasionará em uma crise, mas pode vir a ser uma grande oportunidade que Deus está abrindo em sua vida, como no livro de Números, onde os hebreus foram espionar a terra de Canaã.

## Planejamento e submissão

Muitas vezes precisamos buscar aconselhamentos diretamente a Deus sobre os nossos planos. Às vezes, Ele fala através de outras pessoas ou eventos.

> ²² Quando não há conselhos os planos se dispersam, mas havendo muitos conselheiros eles se firmam. (Provérbios 15:22 ACF)

> ¹⁴ Não havendo sábios conselhos, o povo cai, mas na multidão de conselhos há segurança. (Provérbios 11:14 ACF)

Pessoas sábias abençoam aqueles ao seu redor com bons conselhos.

**Acatar a resposta de Deus ao planejamento é um ato de submissão** do eu à resposta e autoridade divina, independente do que queremos como resposta.

>⁹ O coração do homem planeja o seu caminho, mas o Senhor lhe dirige os passos. (Provérbios 16:9 ACF)

Deus olha diretamente para o nosso coração, ele sabe melhor do que nós do que, quando e como nós precisamos, bem como quais provações devemos passar.

A vida terrena é composta por despesas, viver em sociedade gera despesas que devem ser pagas corretamente, seja a outra pessoa, uma empresa ou o prórpio Estado. Morar, se alimentar, se deslocar, possuir bens. Ter gastos e dívidas são inerentes à vida humana.

Na passagem abaixo, mostra que até Jesus devia impostos. Veja bem, não é o mesmo que inadimplência, quando você deixa vencer a dívida.

> ²⁴ E, chegando eles a Cafarnaum, aproximaram-se de Pedro os que cobravam as dracmas, e disseram: O vosso mestre não paga as dracmas?
²⁵ Disse ele: Sim. E, entrando em casa, Jesus se lhe antecipou, dizendo: Que te parece, Simão? De quem cobram os reis da terra os tributos, ou o censo? Dos seus filhos, ou dos alheios?
²⁶ Disse-lhe Pedro: Dos alheios. Disse-lhe Jesus: Logo, estão livres os filhos.
²⁷ Mas, para que os não escandalizemos, vai ao mar, lança o anzol, tira o primeiro peixe que subir, e abrindo-lhe a boca, encontrarás um estáter; toma-o, e dá-o por mim e por ti. (Mateus 17:24-27 ACF)

Quando provocado, Jesus disse: "Dai a César o que é de César e a Deus, o que é de Deus". Nessa frase dos Evangelho (Mateus 22:21), Ele ensina a distinguir obrigações civis (impostos, leis) da devoção espiritual.

E ainda Paulo disse aos Romanos:

> ⁷ Portanto, dai a cada um o que deveis: a quem tributo, tributo; a quem imposto, imposto; a quem temor, temor; a quem honra, honra.
⁸ A ninguém devais coisa alguma, a não ser o amor com que vos ameis uns aos outros; porque quem ama aos outros cumpriu a lei. (Romanos 13:7-8 ACF)

Gerenciar com sabedoria nossas despesas é um dos pontos chaves para o enriquecimento. Viver abaixo do nosso nível de receita é fundamental:

>²⁰ Há tesouro desejável e azeite na habitação do sábio, mas o homem tolo gasta tudo. (Provérbios 21:20 BKJ)

Não uma questão de ser rico ou pobre, mas de ser sábio ou tolo. O rico herdeiro poderá vir a ser pobre se não for sábio, e vice-versa.

Não é o tamanho do salário, é a adequação das depesas correntes (mensais) ao seu salário.

Se não sobra nada do seu salário após as despesas, significa que você está mantendo um nível de vida maior que deveria, ou que você deveria buscar novas formas de renda para suprir suas necessidades.

As redes sociais são um dreno de tempo e energia. Lá se vive um mundo de fantasia e as pessoas desejam ardentemente ser como as pessoas que eles seguem:

> ¹² Não temos a pretensão de nos igualar ou de nos comparar com alguns que recomendam a si mesmos. Quando eles se medem e se comparam consigo mesmos, agem sem entendimento.
¹³ Nós, porém, não nos gloriaremos além do que é devido, mas nos limitaremos à esfera de ação que Deus nos confiou, a qual alcança também vocês. (2 Coríntios 10:12,13 NVI)

Foque no que reamlmente é valoroso à luz da vida cristã.


## Dívidas e endividamento

Vamos fixar o conceito de endividamento como um estágio mais avançado da dívida devida, pois neste caso a pessoa não tem condições de quitá-la. Quando as pessoas perdem que deveriam administrar o controle financeiro das suas despesas e entram no endividamento.

Deve-se aprender a diferenciar oportunidades de impulsos. O que você comprou lhe renderá melhores condições de prosperar? Ou apenas aumentará suas dívidas para satisfazer seu prazer? A aplicação desse dinheiro te gerará um ativo ou um passivo financeiro?

Se envolver em um endividamento requer bastante estudo e cautela, pois podem se agravar com situações para as quais você não se preparou.

> ²⁶ Não estejas entre os que se comprometem, e entre os que ficam por fiadores de dívidas,
²⁷ Pois se não tens com que pagar, deixarias que te tirassem até a tua cama de debaixo de ti?
²⁸ Não removas os antigos limites que teus pais fizeram. (Provérbios 22:26-28 ACF)

Digamos que uma pessoa quer um aparelho *smartphone* de última linha sem ter recursos para isso. Sem planejamento vai ao banco e pede um empréstimo, ou parcela no cartão.

No mês seguinte, ocorre um sinistro de saúde na família e você terá despesas inesperadas. Como resolver sem uma reserva para emergências, ou sem limite no cartão? Terá que recorrer a outro endividamento de pior qualidade, como o rotativo do cartão, o LIS (limite) da conta bancária, deixando-a no negativo ou ainda pedindo emprestado para uma financeira com juros altíssimos.

Veja que uma decisão errada desencadeou várias outras situações que se agravaram por conta de um ato impulsivo.

### Sobre emprestar o nome (fiador) ou dinheiro

Salomão fala em Provérbios sobre ajudar pessoas sendo-lhe fiador:

> ¹ Filho meu, se ficaste por fiador do teu companheiro, se deste a tua mão ao estranho,
² E te deixaste enredar pelas próprias palavras; e te prendeste nas palavras da tua boca;
³ Faze pois isto agora, filho meu, e livra-te, já que caíste nas mãos do teu companheiro: vai, humilha-te, e importuna o teu companheiro.
⁴ Não dês sono aos teus olhos, nem deixes adormecer as tuas pálpebras.
⁵ Livra-te, como a gazela da mão do caçador, e como a ave da mão do passarinheiro. (Provérbios 6:1-5 ACF)

Pensando nos dias de hoje, se uma pessoa precisa de dinheiro, o caminho mais correto é pedir a uma instituição que tem essa finalidade.

Quando comecei a trabalhar, li em um livro a seguinte frase: *antes de pedir dinheiro a um amigo, pense de qual dos dois você precisa mais*. Um relacionamento que envolve dinheiro pode ser arruinada com muita facilidade. O que pediu emprestado fica desconfiado, se afasta, sente-se vigiado pelos seus gastos. O relacionamento se verticaliza, ou seja, o fiador é colocado pelo outro como uma pessoa superior, a quem ele deve prestar contas.

A Bíblia fala que devemos evitar ser fiador ou de avalizar em outros versículos. Você compromente sua honra pelo compromisso de outro.

>¹ Filho meu, se ficaste por fiador do teu companheiro
  e se te empenhaste ao estranho,
² estás enredado com o que dizem os teus lábios,
  estás preso com as palavras da tua boca.
³ Agora, pois, faze isto, filho meu,
  e livra-te, pois caíste nas mãos do teu companheiro:
  vai, prostra-te e importuna o teu companheiro;
⁴ não dês sono aos teus olhos,
  nem repouso às tuas pálpebras;
⁵ livra-te, como a gazela, da mão do caçador
  e, como a ave, da mão do passarinheiro (Provérbios 6:1-5 ARA)

Isso inclui emprestar seu nome ou seu cartão. Se a pessoa não teve zelo pelo próprio nome para possuir "crédito na praça", imagine o que ela poderá fazer com o nome de outrém.

>¹⁶ Tome-se a roupa àquele que fica fiador por outrem;
  e, por penhor, àquele que se obriga por estrangeiros. (Provérbios 20:16 ARA)

> ¹⁵ Quem serve de fiador a um desconhecido certamente sofrerá, mas quem se nega a comprometer-se está seguro. (Provérbios 11:15 NVI)

Portanto, ao adquirir dívidas você deve fazer de forma calculada, com planejamento, e com a direção de Deus.

As lições que eu aprendi vivenciando as relações com familiares e amigos são:

- Não se deve emprestar dinheiro nem a parente, imagine a estranhos.

- Nunca empreste um dinheiro que você precisa. Na pior das hipóteses, tenha uma reserva de empréstimos que você não tenha necessidade de receber de volta.

- Emprestar dinheiro é uma forma de dar corda para uma pessoa se enforcar. Se puder não emprestar, ao invés dê conselhos, pois o dinheiro só vai acomodar a outra pessoa. Com conselhos ela pode tentar formas criativas de ganhar um dinheiro extra.

- Se você emprestou dinheiro e não quer perder o relacionamento, perdoe a dívida. É um fardo a menos.

- Emprestar **afeta** relacionamentos, as pessoas nunca mais serão as mesmas quando se encararem.

## Prejuízos financeiros e espirituais da dívida

A Bíblia nos alerta em Provérbios como uma dívida pode arruinar sua vida e a de sua família:

> O rico domina sobre o pobre; quem toma emprestado é escravo de quem empresta. (Provérbios 22:7 NVI)

Escravo bem do hebraico *eved*, substantivo do verbo *avad* que significa trabalhar ou servir. O mesmo termo utilizado sobre os escravos hebreus no Egito.

Permanecer endividado leva a pessoa a se tornar uma bola de neve e a sua vida vai mudando se você inconsientemente. Você passa a evitar pessoas e lugares. Pode escandalizar, virar alvo de comentários. A reputação se corrói. E isso se torna uma consequência.

Boa parte dos fins de relacionamentos são causados de **traição financeira**.

> ²⁴ As mãos diligentes governarão, mas os preguiçosos acabarão em trabalho forçado. (Provérbios 12:24 NVI)

Seu salário passa a servir para pagar os juros da dívida, você perde qualidade de vida e paz, perde seu tempo para recompensar o diligente.

>²⁶ Não estejas entre os que se comprometem
  e ficam por fiadores de dívidas,
²⁷ pois, se não tens com que pagar,
  por que arriscas perder a cama de debaixo de ti (Provérbios 22:26,27 ARA)

Na pior das hipóteses, sua dívida pode ser cobrada e os bens que você construiu retirados de si.

## Credor

Se você já emprestou, melhor que tenha dado e que não conte com o dinheiro de volta. Quando uma pessoa pede dinheiro emprestado:

1. Foi uma emergência inesperada acima de suas reservas de emergência;
2. A pessoa precisa de recursos para realizar um projeto e pretende multiplicar o dinheiro;
3. Ela não se organizou financeiramente para emergências;
4. Ela vive num padrão acima da sua renda.
5. Não possui certidão positiva (leia-se nome limpo) para solicitar um empréstimo formalmente para uma instituição financeira.

Em qualquer hipótese, ao emprestar não espere que você receba de volta. Contabilize como uma bênção o retorno, pois se envolver nessa situação pode perturbar a sua paz.

## Proposta lúdica 1 (Power Grid)

O Power Grid é um dos melhores simuladores de negócios disfarçados de jogo de tabuleiro. Como o núcleo do jogo é gerenciar recursos escassos para expandir uma rede de energia, ele é uma ferramenta excelente para ilustrar conceitos reais de gestão financeira, especialmente o controle e a estratégia de despesas.

A dinâmica familiar lida com os mesmos desafios de fluxo de caixa, controle de impulsos e custos ocultos que uma empresa de energia.

Neste jogo onde você deve planejar suas despesas para conseguir suprir todas as etapas da rodada:
- **leilão de plantas**: lidar ocm o impulso de comprar;
- **compra de insumos**: lidar com inflação e sazonalidade de valores;
- **expansão da rede de firnecimento**: avaliação de quando é a hora de investir.




## Proposta lúdica 2 (Railways of the World)

Railways of the World é um jogo brilhante porque, diferentemente de outros jogos onde você já começa com um bom capital, nele você é forçado a lidar com o peso do crédito desde o primeiro turno. Ele é um verdadeiro simulador de alavancagem financeira e planejamento de rotas.

Trazer essa dinâmica para a gestão de despesas familiares ajuda a visualizar como o crédito, o planejamento e os custos ocultos afetam a "trilha de pontuação" do seu orçamento doméstico. Aqui estão as principais formas de associar as mecânicas do jogo com as finanças da sua casa:
- **Títulos da dívida (bonds)**: metáfora para o uso do cartão de crédito, cheque especial e financiamentos;
- ** Custo dos terrenos**: reflete seu estilo de vida, deve-se avaliar se faz o caminho mais curto e às vezes mais caro.

## Diversificação de Investimentos
Em Eclesiastes, fala sobre a não aplicação, mas é são palavras de sabedoria aplicáveis em qualquer esfera. Diversificar é se proteger.

> Reparta o que você tem com sete, até mesmo com oito, pois você não sabe que desgraça poderá cair sobre a terra. (Eclesiastes 11:2 NVI)

# 5 - Propósito

O propósito é o combustível para você sobrepor os obstáculos, as dificuldades. O que te motiva a trabalhar, economizar e investir?

>²⁶ O apetite do trabalhador o obriga a trabalhar; a sua fome o impulsiona. (Provérbios 16:26 NVI)

Qual é a sua fome?

É o status? É mostrar para as pessoas? Ou para melhorar a qualidade de vida de sua família?

Dinheiro é uma ferramenta, não a finalidade. É uma bênção que permite abençoar outras pessoas.

A pessoa que prioriza o dinheiro ela troca sua vida (tempo, saúde, relacionamento, paz e sono), e por que não dizer sua alma, por ele. No dinheiro estará seu coração e ele será o motivo da sua perdição.


Receber dinheiro não é o problema, mas sim a motivação que te levam a ganhá-lo:

>⁹ Os que querem ficar ricos caem em tentação, em armadilhas e em muitos desejos descontrolados e nocivos, que levam os homens a mergulharem na ruína e na destruição,
¹⁰ pois o amor ao dinheiro é raiz de todos os males. Algumas pessoas, por cobiçarem o dinheiro, desviaram-se da fé e se atormentaram a si mesmas com muitos sofrimentos. (1 Timóteo 6:10 NVI)

## Caridade e generosidade

O equilíbrio entre caridade e crescimento deixa um legado duradouro. Riqueza com propósito garante provisão não só para sua família, mas para outras pessoas.

> ³⁸ Dai, e ser-vos-á dado; boa medida, recalcada, sacudida e transbordando, vos deitarão no vosso regaço; porque com a mesma medida com que medirdes também vos medirão de novo. (Lucas 6:38 ACF)

Não é uma troca, é a multiplicação de dos princípios cristãos com seus recursos, seu dinheiro, seu tempo. É uma postura, um plantio. Você abre mão de um pouco para semear prosperidade aos próximos.

**Somos abençoados a abençoar outras pessoas.**

>²⁸ Aquele que furtava não furte mais; antes, trabalhe, fazendo com suas mãos aquilo que é bom, **para que tenha o que dar ao que tiver necessidade**. (Efésios 4:28 BKJ)

Contribua conforme o coração, não doe com tristeza, com soberba para alimentar seu ego nem com intenções de retorno.

>⁷ Cada um contribua conforme determinou no coração, não com pesar nem por obrigação, pois Deus ama a quem dá com alegria.
⁸ Deus é poderoso para fazer que toda a graça seja acrescentada a vocês, para que em todas as coisas, em todo o tempo, tendo tudo o que é necessário, vocês transbordem em toda boa obra.
⁹ Como está escrito: "Distribuiu generosamente aos pobres; a sua justiça permanece para sempre". (2 Coríntios 9:7-9 NVI)

Muito cuidado com emprestar dinheiro, prefira doá-lo. Se possível, de um caixa separado para doações.

Isso causará menos problemas entre você e a outra pessoa. Se o dinheiro retornar, será uma bênção. Se não retornar, não deve lhe fazer falta.
