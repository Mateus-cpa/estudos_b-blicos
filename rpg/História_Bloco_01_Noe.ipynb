# As Sete Portas do Recomeço

A campanha acompanha a queda, o primeiro juízo, a preservação de Noé e o recomeço da humanidade, com foco em pecado, arrependimento, justiça, provisão e obediência.

O tom é de drama fundacional: os jogadores não “substituem” os personagens bíblicos, mas ajudam a proteger, advertir, preparar e sustentar aquilo que já está determinado na narrativa.


# Premissa
Os jogadores representam uma pequena equipe de servos, parentes, artesãos, mensageiros ou conselheiros ligados à linhagem de Noé antes do dilúvio.

Eles percebem sinais de corrupção crescente, ajudam a reunir recursos para a arca, protegem a família de Noé de sabotagem e enfrentam a hostilidade social que cresce à medida que o mundo rejeita o caminho de Deus.

Cada missão bem-sucedida não altera a Bíblia; ela apenas garante que Noé consiga cumprir aquilo que já foi ordenado.


## O Antagonista: Malak, o Escarnecedor e Arquiteto da Corrupção

**Nome:** Malak (que significa 'mensageiro' ou 'anjo' em hebraico, mas com um significado invertido, representando um "mensageiro da perdição" ou "anjo caído" figurativo).

**Perfil:** Malak é um líder carismático e manipulador, influente na sociedade pré-diluviana. Ele representa a voz da razão mundana, do hedonismo e da negação da corrupção espiritual. Ele não é necessariamente "mau" no sentido demoníaco, mas personifica a rebelião humana e a rejeição ao caminho de Deus.

**Motivações:**
*   **Poder e Influência:** Malak prospera no caos e na depravação da sociedade. A ordem e a moralidade que Noé prega ameaçam seu status e sua forma de vida.
*   **Ceticismo e Escárnio:** Ele vê Noé e sua família como tolos fanáticos, e a ideia de um dilúvio como absurda. Ele usa o ridículo e a lógica deturpada para minar a fé e a credibilidade de Noé.
*   **Manutenção do Status Quo:** Malak representa aqueles que se beneficiam da corrupção e se recusam a reconhecer a necessidade de mudança ou juízo.

**Táticas:**
*   **Guerra Psicológica:** Malak não usa a força bruta diretamente (embora possa incitar outros a fazê-lo). Ele usa boatos, calúnias, argumentos "racionais" distorcidos e zombaria pública para isolar Noé e seus aliados.
*   **Sabotagem Indireta:** Pode organizar o roubo de materiais, espalhar desinformação sobre a arca ou até mesmo tentar aliciar membros da equipe de Noé com promessas de riqueza ou prazer mundano.
*   **Liderança de Oposição:** Ele seria o principal instigador da hostilidade social mencionada na premissa, organizando manifestações, boicotes e ataques verbais contra a construção da arca.

**Impacto na Trama:**
*   **Ato 1 (O mundo endurece):** Malak serve como a personificação da **"dureza do mundo"**, liderando o escárnio e a violência moral contra Noé e qualquer um que o ouça. Ele organiza reuniões públicas onde as advertências são ridicularizadas e transforma a passividade em oposição ativa.
*   **Ato 2 (O chamado de Noé):** Tenta ativamente desacreditar a revelação divina de Noé, plantando sementes de dúvida sobre sua sanidade e tentando destruir ou roubar os planos e projetos iniciais da arca, impedindo o progresso desde o início.
*   **Ato 3 (A sabotagem):** Ele é o arquiteto da **"sabotagem"**, coordenando esforços para atrasar a construção da arca. Isso inclui incitar trabalhadores a abandonar o projeto, organizar roubos de materiais essenciais e espalhar rumores sobre a má qualidade ou instabilidade da construção.
*   **Ato 4 (A última advertência):** Sua influência torna a **"última advertência"** mais difícil, pois ele terá envenenado a mente das pessoas contra a mensagem de Noé. Ele organiza distrações grandiosas e festividades para abafar as últimas chamadas ao arrependimento, assegurando que a mensagem não alcance corações abertos.
*   **Ato 5 (A porta se fecha):** No clímax, ele lidera uma turba desesperada para impedir a entrada na arca, seja para forçar sua própria entrada, para destruí-la em um ataque de raiva, ou simplesmente para impedir que Noé e sua família a ocupem, usando seu carisma para incitar a multidão ao pânico e à violência.
*   **Para os jogadores:** Malak representará um desafio constante à sua vontade, sua fé e sua capacidade de cumprir a missão. Derrotá-lo (não necessariamente fisicamente, mas minando sua influência ou expondo sua hipocrisia) seria uma vitória moral significativa, embora ele provavelmente pereça no dilúvio como um símbolo da humanidade que recusou a salvação.

### Ato 1 - Cena 1: A Corrupção Silenciosa e os Sinais Iniciais
Os jogadores, em suas rotinas diárias (colheita, comércio, viagens), começam a notar pequenas anomalias. Notícias de vilas saqueadas por bandos sem lei, mercadores agindo com desonestidade flagrante, ou pessoas se entregando a prazeres excessivos sem pudor. O comportamento humano degenera rapidamente.

Enquanto viajam entre aldeias para buscar suprimentos, os personagens encontram um grupo de fazendeiros carregando bens pessoais.

*   **Teste Principal:** `Percepção` (para notar os detalhes do saque e o desespero dos fazendeiros ou para sentir a gravidade da situação e o que ela representa).
    *   **Sucesso:**
        *   **1 Símbolo de Fé:** Os jogadores compreendem a escala da violência e a desolação, ganhando clareza moral sobre o estado do mundo. (+1 Símbolo de Fé: clareza moral).
        *   **2 Símbolos de Fé:** Eles percebem que este não é um incidente isolado, mas parte de um padrão crescente, identificando o perigo cedo. (+2 Símbolos de Fé: alerta útil).
    *   **Falha:**
        *   **1 Complicação:** O grupo demora a perceber a gravidade da situação, talvez a tratando como um incidente isolado de banditismo, perdendo a oportunidade de investigar mais a fundo. (Complicação 1: atraso na compreensão).


Na verdade, estão fugindo de sua vila que foi saqueada por 'homens de renome' que ninguém ousou desafiar. Os poucos sobreviventes estão aterrorizados e sem esperança.



### Ato 1 - Cena 2: O Silêncio dos Justos e a Festa do Abismo

Ouvem-se entre os aldeões sobre ruídos de festas noturnas que se estendem por dias, ausência de anciãos respeitados em conselhos, jovens desrespeitando tradições ancestrais.

E o pior, que os próprios homens de Deus estão se corrompendo.

**Situação:** Enquanto os jogadores retornam ou circulam por áreas antes tranquilas, eles percebem que o mal não está apenas nos 'homens de renome' distantes, mas entranhado no cotidiano. Anciãos respeitados agora fecham os olhos para injustiças em troca de prazeres, e a juventude celebra a rebeldia como virtude.

**Cenário:** O entardecer em uma vila ou bairro periférico. Uma festa ruidosa e caótica bloqueia o caminho principal. O que deveria ser uma celebração de colheita tornou-se um evento de excessos, onde o desrespeito aos pais e aos costumes sagrados é a regra.

*   **Teste Principal:** `Intuição` ou `Teologia` (para discernir a profundidade da corrupção espiritual e não ser tragado pelo clima de permissividade).
    *   **Sucesso:**
        *   **1 Símbolo de Fé:** **Clareza Moral.** O jogador percebe exatamente onde a semente da corrupção brotou naquele grupo, evitando ser enganado por falsas aparências de 'alegria'.
        *   **2 Símbolos de Fé:** **Exemplo Silencioso.** A postura sóbria e digna do jogador causa um breve momento de hesitação ou vergonha em um conhecido que estava se perdendo na multidão (+2 Símbolos de Fé: um contato útil é lembrado de sua ética).
        *   **3 Símbolos de Fé:** **Leitura Profética.** O jogador antecipa um conflito violento que está prestes a explodir na festa e consegue retirar seu grupo ou um inocente antes que o sangue seja derramado.
    *   **Falha:**
        *   **1 Complicação:** **Desgaste Emocional.** O cinismo dos aldeões e as ofensas proferidas contra os valores dos jogadores pesam no espírito, causando um redutor temporário em testes de Vontade na próxima cena.
        *   **2 Complicações:** **Descrédito Social.** Ao tentarem intervir ou apenas passar, os jogadores são marcados como 'puritanos' ou 'estranhos', dificultando diálogos futuros com a liderança local.

**Dica para o Narrador:**
Descreva os sons: o riso que soa forçado e histérico, o som de jarros de barro quebrando e o silêncio desconfortável dos poucos anciãos que restam, sentados nas sombras, com medo de falar. O objetivo é mostrar que a lei de Deus foi substituída pela vontade do mais forte ou do mais ruidoso.

### Ato 2 - Cena 1: O Chamado e o Peso da Obediência

**Situação:** Após a revelação divina a Noé, a família e seus servos próximos se reúnem. O choque da notícia do Dilúvio é acompanhado pela magnitude da tarefa: construir uma embarcação colossal em um mundo que nunca viu chuva. Os jogadores devem ajudar a traduzir a visão espiritual em um plano de logística e construção, enquanto lidam com suas próprias dúvidas.

**Cenário:** O acampamento de Noé, ao amanhecer. Noé traça no chão as dimensões da Arca. Há um silêncio pesado; o ar parece denso com a importância do momento.

*   **Teste Principal:** `Teologia` (para discernir a vontade divina nos detalhes) ou `Ofício` (para planejar a execução técnica inicial).
    *   **Sucesso:**
        *   **1 Símbolo de Fé:** **Direção Clara.** Os jogadores identificam a melhor fonte de madeira de gofer na região, economizando tempo precioso na mobilização.
        *   **2 Símbolos de Fé:** **Firmeza de Espírito.** A confiança dos jogadores ao organizar os planos acalma os membros mais jovens da família, reduzindo o medo inicial. (+2 Símbolos de Fé: bônus em testes de liderança nesta cena).
        *   **3 Símbolos de Fé:** **Confirmação do Propósito.** Durante o planejamento, um sinal sutil (como a harmonia incomum dos animais ao redor) confirma que a mão de Deus está sobre o projeto, renovando a esperança do grupo.
    *   **Falha:**
        *   **1 Complicação:** **Dúvida/Hesitação.** Erros no planejamento inicial causam desperdício de ferramentas ou confusão na atribuição de tarefas. (Atraso na montagem do canteiro de obras).
        *   **2 Complicações:** **Peso Emocional.** A enormidade da tarefa sobrecarrega um dos jogadores, gerando uma crise de ansiedade que exige um teste de Vontade para não perder o foco nos próximos dias.

**Dica para o Narrador:**
Enfatize o tamanho. Use objetos do cotidiano dos jogadores para comparar: 'A Arca terá o comprimento de dez caravanas enfileiradas'. Descreva Noé não como um general, mas como um homem que carrega um fardo sagrado, cujos olhos refletem tanto o temor a Deus quanto a tristeza pelo que virá. O objetivo é fazer os jogadores sentirem que o trabalho manual que eles farão é, em si, um ato de adoração e sobrevivência.

### Ato 2 - Cena 2: O Boato e o Escárnio Velado

**Situação:** As notícias sobre a construção da Arca e as pregações de Noé começam a circular não como um aviso sério, mas como a piada do século. Nos mercados e poços, o nome de Noé é associado à loucura. Agentes de Malak espalham rumores de que a família de Noé está estocando recursos para um golpe ou que enlouqueceram por visões febris.

**Cenário:** O mercado central da cidade. Os jogadores precisam obter suprimentos específicos (como betume ou ferramentas de bronze) que se tornaram subitamente difíceis de comprar devido ao estigma social.

*   **Teste Principal:** `Diplomacia` ou `Vontade` (para navegar na hostilidade social, obter os recursos necessários sem ceder à raiva ou ao desânimo).
    *   **Sucesso:**
        *   **1 Símbolo de Fé:** **Paz Interior.** O grupo ignora os insultos, mantendo o foco. Os mercadores vendem o básico por um preço justo apenas para se livrarem deles rapidamente.
        *   **2 Símbolos de Fé:** **Palavra Certa.** Um dos jogadores responde a um insulto com tamanha sabedoria que silencia a roda de escarnecedores momentaneamente, ganhando o respeito silencioso de um fornecedor que oferece material de melhor qualidade. (+2 Símbolos de Fé: recurso extra ou desconto).
        *   **3 Símbolos de Fé:** **Desarma Conflito.** A postura do grupo é tão íntegra que uma pequena parte da multidão começa a questionar se Noé não teria razão, criando uma brecha na influência de Malak naquela área.
    *   **Falha:**
        *   **1 Complicação:** **Vergonha/Ostracismo.** O grupo é expulso de um estabelecimento sob risos, forçando-os a procurar em locais mais perigosos ou distantes. (Atraso na coleta).
        *   **2 Complicações:** **Provocação.** Um grupo de jovens incitados por Malak começa a empurrar os jogadores ou danificar suas mercadorias, exigindo um teste de Luta ou Fuga para evitar ferimentos leves.

**Dica para o Narrador:**
Use Malak não como um vilão que ataca, mas como o 'comentarista' irônico. Ele pode aparecer ao fundo, cercado de seguidores, apenas observando e apontando para os jogadores enquanto faz uma piada sobre 'barcos no deserto', provocando gargalhadas na multidão. O objetivo é fazer os jogadores sentirem o peso de serem 'os loucos' aos olhos da sociedade.

### Ato 2 - Cena 3: Demarcação do Local e Preparação da Obra

**Situação:** O projeto sai do papel. Noé e sua família começam a limpar o terreno e a cravar as estacas que definem o perímetro colossal da Arca. Malak, percebendo que as palavras não detiveram o grupo, intensifica a oposição, organizando a 'Festa do Escárnio' nas proximidades para distrair trabalhadores e intimidar fornecedores.

**Cenário:** Um descampado elevado. O sol castiga o solo seco enquanto os jogadores transportam as primeiras vigas de sustentação e ferramentas pesadas sob o olhar vigilante e hostil da multidão liderada por Malak.

*   **Teste Principal:** `Ofício` (para a precisão da demarcação e montagem da infraestrutura) ou `Sobrevivência` (para gerenciar a logística de suprimentos sob boicote).
    *   **Sucesso:**
        *   **1 Símbolo de Fé:** **Fundação Sólida.** O trabalho inicial é feito com tamanha excelência que a estrutura suportará pressões futuras sem ceder. (+1 Símbolo de Fé: estabilidade estrutural).
        *   **2 Símbolos de Fé:** **Provisão Inesperada.** Apesar do boicote de Malak, um mercador estrangeiro, não influenciado pelos boatos locais, chega com ferramentas de alta qualidade e aceita negociar. (+2 Símbolos de Fé: recurso decisivo).
        *   **3 Símbolos de Fé:** **Autoridade Moral.** A dedicação incansável dos jogadores diante do sol e dos insultos começa a silenciar os críticos mais moderados da multidão, mudando a atitude de alguns observadores.
    *   **Falha:**
        *   **1 Complicação:** **Ferramenta Quebrada.** O desgaste físico e a pressa resultam na perda de equipamentos essenciais, exigindo improvisação difícil.
        *   **2 Complicações:** **Sabotagem Parcial.** Durante a noite ou um momento de distração, agentes de Malak removem estacas de demarcação ou contaminam a água do acampamento, causando atrasos e fadiga.

**Dica para o Narrador:**
Descreva o contraste visual: de um lado, o trabalho ordeiro, suado e sagrado da família de Noé; do outro, a música alta, o deboche e a embriaguez da 'Festa do Escárnio' de Malak. O ar deve parecer eletrizante, como se a própria terra estivesse desconfortável com a presença da Arca. Malak deve ser visto brindando 'à saúde dos que constroem navios na poeira', enquanto a multidão explode em gargalhadas.

## Ato 3 — A sabotagem
Enquanto a arca é construída, surgem pressões externas na tentativa de impedir a construção: zombaria, roubo de materiais, incêndio da floresta.

Esse trecho ensina perseverança, honestidade e fidelidade no trabalho diário.

**Malak:**
No Ato 3, ele é o arquiteto da **"sabotagem"**, coordenando esforços para atrasar a construção da arca.


### Ato 3 - Cena 1: O Cerco aos Suprimentos

**Situação:** Malak organiza milícias locais para bloquear as rotas de comércio e interceptar as caravanas que trazem betume e madeira de gofer. Ele não ataca a Arca diretamente ainda, mas tenta 'sufocar' a obra retirando seus recursos.

**Cenário:** Uma ravina estreita ou uma estrada poeirenta onde uma carroça de suprimentos foi emboscada. Os jogadores chegam e encontram os condutores acuados por capangas de Malak.

*   **Teste Principal:** `Luta` ou `Logística/Sobrevivência` (para recuperar os bens ou encontrar uma rota alternativa segura).
    *   **Sucesso:**
        *   **1 Símbolo de Fé:** **Recuperação.** Os suprimentos são salvos com danos mínimos.
        *   **2 Símbolos de Fé:** **Rota Oculta.** O grupo descobre um caminho antigo e esquecido que permite contornar os bloqueios de Malak no futuro. (+2 Símbolos de Fé: bônus em futuras coletas).
    *   **Falha:**
        *   **1 Complicação:** **Escassez.** Metade do betume é derramado ou roubado, exigindo que a Arca seja construída mais lentamente.
        *   **2 Complicações:** **Ferimento.** Um aliado importante é ferido, exigindo cuidados e reduzindo a mão de obra disponível.

### Ato 3 - Cena 2: O Fogo na floresta

**Situação:** Desesperado com o progresso, Malak ordena que seus seguidores ateiem fogo na floresta de onde a madeira está sendo extraída. O objetivo é destruir a matéria-prima e criar um desastre ambiental que distraia os construtores.

**Cenário:** A borda da floresta durante a noite. Chamas alaranjadas sobem aos céus e o vento sopra o calor em direção ao canteiro de obras da Arca.

*   **Teste Principal:** `Resistência` ou `Atletismo` (para conter o fogo e resgatar ferramentas ou animais antes que as chamas se espalhem).
    *   **Sucesso:**
        *   **1 Símbolo de Fé:** **Proteção do Espólio.** O fogo é contido antes de atingir o estoque principal de madeira cortada.
        *   **2 Símbolos de Fé:** **Chuva Providencial.** Uma neblina densa ou orvalho incomum surge, ajudando a apagar as brasas mais perigosas. (+2 Símbolos de Fé: alívio da fadiga).
    *   **Falha:**
        *   **1 Complicação:** **Perda de Equipamento.** O fogo consome o depósito de ferramentas de bronze, tornando os testes de 'Ofício' mais difíceis no próximo ato.
        *   **2 Complicações:** **Inalação de Fumaça.** O grupo sofre penalidades físicas devido ao esforço extremo e à fumaça tóxica.
        * **3 Complicações:**
        ** Atraso no retorno e roubo dos suprimentos**: O grupo demora tanto tempo para apagar o incêndio que a construção fica exposta e é saqueada.

### Ato 3 - Cena 3: Invasão e Sabotagem Silenciosa

**Situação:** Sob o manto de uma tempestade de areia (ou escuridão profunda), infiltrados de Malak tentam entrar no canteiro de obras para serrar suportes estruturais ou envenenar os animais que já começaram a se aproximar.

**Cenário:** O esqueleto da Arca, uma catedral de madeira inacabada. Sombras se movem entre as vigas enquanto o vento uiva.

*   **Teste Principal:** `Guardião` ou `Percepção` (para detectar os intrusos e proteger a integridade da estrutura).
    *   **Sucesso:**
        *   **1 Símbolo de Fé:** **Vigilância.** Os sabotadores são expulsos antes de causarem qualquer dano estrutural.
        *   **2 Símbolos de Fé:** **Justiça Exposta.** Um dos sabotadores é capturado e, ao ser confrontado, revela os planos de Malak para o próximo ato. (+2 Símbolos de Fé: informação estratégica).
    *   **Falha:**
        *   **1 Complicação:** **Dano Estrutural.** Uma viga mestre é comprometida, exigindo um reparo de emergência perigoso.
        *   **2 Complicações:** **Invasão do Perímetro.** A sensação de segurança é quebrada; o medo se instala entre a família, exigindo testes de Vontade constantes.

**Dica para o Narrador do Ato 3:**
Aumente a tensão física. O perigo agora é real e pode causar morte ou ferimentos graves. Mostre que a Arca, embora seja um projeto de Deus, exige o suor e o sangue dos homens para ser defendida contra o mal que não quer ser julgado.

## Ato 4 — A última advertência
Com o tempo se esgotando, os personagens precisam garantir que a mensagem de advertência chegue a pessoas-chave e que a família de Noé permaneça unida.

A lição aqui é que Deus adverte antes de julgar, e que desprezar o aviso tem consequências reais. Os testes decisivos são de Mensageiro, Diplomacia, Teologia e Vontade.

**Malak:**
*   Sua influência torna a **"última advertência"** (Ato 4) mais difícil, pois ele terá envenenado a mente das pessoas contra a mensagem de Noé.

### Ato 4 - Cena 1: Cicatrizes na Madeira e no Espírito (Consequências da Sabotagem)

**Situação:** A Arca permanece, mas o canteiro de obras é um cenário de desolação. Vigas chamuscadas pelo fogo, ferramentas perdidas e o cansaço extremo marcam o rosto de cada familiar. Malak recuou fisicamente, mas sua vitória é o silêncio hostil da vizinhança: ninguém mais ousa olhar para Noé, e o medo de novos ataques paralisa os ajudantes restantes. Os jogadores precisam lidar com os danos e decidir quem ainda merece ouvir a última advertência.

**Cenário:** O pátio interno da Arca inacabada, sob um céu estranhamente opaco. O cheiro de cinzas ainda paira no ar. Noé está em silêncio, observando o horizonte.

*   **Teste Principal:** `Ofício` (para reparos emergenciais) ou `Vontade/Diplomacia` (para reerguer o moral da família e dos últimos servos).
    *   **Sucesso:**
        *   **1 Símbolo de Fé:** **Restauração.** Um reparo crítico é feito com perfeição, devolvendo a sensação de segurança estrutural à Arca.
        *   **2 Símbolos de Fé:** **Unidade Inabalável.** Um discurso ou ato de serviço dos jogadores remove o medo do coração de um membro da família, garantindo que ninguém desista agora. (+2 Símbolos de Fé: bônus em testes de grupo).
        *   **3 Símbolos de Fé:** **O Remanescente.** Um antigo aliado que havia fugido durante o incêndio retorna, arrependido, trazendo consigo uma informação vital sobre os próximos passos de Malak ou um recurso escondido.
    *   **Falha:**
        *   **1 Complicação:** **Fadiga Crônica.** O esforço para consertar o que foi destruído deixa os jogadores exaustos (Penalidade em testes físicos na próxima cena).
        *   **2 Complicações:** **Desalento.** A notícia de que certos materiais não podem mais ser repostos se espalha, criando um clima de 'missão impossível' entre os construtores.
        *   **3 Complicações:** **Infecção de Dúvida.** A sabotagem de Malak plantou uma desconfiança interna; discussões começam a surgir sobre se a Arca realmente resistirá ao que Noé profetiza.

**Dica para o Narrador:**
Esta é uma cena de 'pé no chão'. Foque nos detalhes das mãos calejadas, nas ferramentas cegas e no vazio deixado por quem partiu. O objetivo é mostrar que a obediência a Deus tem um custo social e físico, preparando o terreno para a urgência da *Última Advertência*. Malak não precisa aparecer; sua presença é sentida no vácuo de ajuda e na hostilidade silenciosa da cidade ao longe.

## Ato 5 — A porta se fecha
O clímax ocorre quando o grupo precisa completar as últimas providências da arca e reagir ao momento em que o juízo começa. O tema é obediência tardia, perda de oportunidade e a seriedade do tempo de Deus.

Aqui entram testes de Resistência, Vontade, Guardião e possivelmente Sacerdote, se houver personagens com esse papel.

> ¹ Depois disse o Senhor a Noé: Entra tu e toda a tua casa na arca, porque tenho visto que és justo diante de mim nesta geração.

> ² De todos os animais limpos tomarás para ti sete e sete, o macho e sua fêmea; mas dos animais que não são limpos, dois, o macho e sua fêmea.

> ³ Também das aves dos céus sete e sete, macho e fêmea, para conservar em vida sua espécie sobre a face de toda a terra.

> ⁴ Porque, passados ainda sete dias, farei chover sobre a terra quarenta dias e quarenta noites; e desfarei de sobre a face da terra toda a substância que fiz.

> ⁵ E fez Noé conforme a tudo o que o Senhor lhe ordenara.

> ⁶ E era Noé da idade de seiscentos anos, quando o dilúvio das águas veio sobre a terra.

> ⁷ Noé entrou na arca, e com ele seus filhos, sua mulher e as mulheres de seus filhos, por causa das águas do dilúvio.

> ⁸ Dos animais limpos e dos animais que não são limpos, e das aves, e de todo o réptil sobre a terra,

> ⁹ Entraram de dois em dois para junto de Noé na arca, macho e fêmea, como Deus ordenara a Noé.

> ¹⁰ E aconteceu que passados sete dias, vieram sobre a terra as águas do dilúvio. (Gênesis 7:1-10, ACF)

**Malak:**
*   No clímax do Ato 5, ele pode estar entre os que tentam impedir a entrada na arca, usando seu carisma para incitar a multidão.


## Ato 6 - O dilúvio

> ¹¹ No ano seiscentos da vida de Noé, no mês segundo, aos dezessete dias do mês, naquele mesmo dia se romperam todas as fontes do grande abismo, e as janelas dos céus se abriram,

> ¹² E houve chuva sobre a terra quarenta dias e quarenta noites.

> ¹³ E no mesmo dia entraram na arca Noé, seus filhos Sem, Cão e Jafé, sua mulher e as três mulheres de seus filhos.

> ¹⁴ Eles, e todo o animal conforme a sua espécie, e todo o gado conforme a sua espécie, e todo o réptil que se arrasta sobre a terra conforme a sua espécie, e toda a ave conforme a sua espécie, pássaros de toda qualidade.

> ¹⁵ E de toda a carne, em que havia espírito de vida, entraram de dois em dois para junto de Noé na arca.

> ¹⁶ E os que entraram eram macho e fêmea de toda a carne, como Deus lhe tinha ordenado; e o Senhor o fechou dentro.

> ¹⁷ E durou o dilúvio quarenta dias sobre a terra, e cresceram as águas e levantaram a arca, e ela se elevou sobre a terra.

> ¹⁸ E prevaleceram as águas e cresceram grandemente sobre a terra; e a arca andava sobre a face das águas.

> ¹⁹ E as águas prevaleceram excessivamente sobre a terra; e todos os altos montes que havia debaixo de todo o céu, foram cobertos.

> ²⁰ Quinze côvados acima prevaleceram as águas; e os montes foram cobertos.

> ²¹ E expirou toda a carne que se movia sobre a terra, tanto de ave como de gado e de feras, e de todo o réptil que se arrasta sobre a terra, e todo o homem.

> ²² Tudo o que tinha fôlego de espírito de vida em suas narinas, tudo o que havia em terra seca, morreu.

> ²³ Assim foi destruído todo o ser vivente que havia sobre a face da terra, desde o homem até ao animal, até ao réptil, e até à ave dos céus; e foram extintos da terra; e ficou somente Noé, e os que com ele estavam na arca.

> ²⁴ E prevaleceram as águas sobre a terra cento e cinquenta dias. (Gênesis 7:11-24, ACF)

## Ato 7 - Em busca de terras


> ¹ E lembrou-se Deus de Noé, e de todos os seres viventes, e de todo o gado que estavam com ele na arca; e Deus fez passar um vento sobre a terra, e aquietaram-se as águas.

> ² Cerraram-se também as fontes do abismo e as janelas dos céus, e a chuva dos céus deteve-se.

> ³ E as águas iam-se escoando continuamente de sobre a terra, e ao fim de cento e cinquenta dias minguaram.

> ⁴ E a arca repousou no sétimo mês, no dia dezessete do mês, sobre os montes de Ararate.

> ⁵ E foram as águas indo e minguando até ao décimo mês; no décimo mês, no primeiro dia do mês, apareceram os cumes dos montes.

> ⁶ E aconteceu que ao cabo de quarenta dias, abriu Noé a janela da arca que tinha feito.

> ⁷ E soltou um corvo, que saiu, indo e voltando, até que as águas se secaram de sobre a terra.

> ⁸ Depois soltou uma pomba, para ver se as águas tinham minguado de sobre a face da terra.

> ⁹ A pomba, porém, não achou repouso para a planta do seu pé, e voltou a ele para a arca; porque as águas estavam sobre a face de toda a terra; e ele estendeu a sua mão, e tomou-a, e recolheu-a consigo na arca.

> ¹⁰ E esperou ainda outros sete dias, e tornou a enviar a pomba fora da arca.

> ¹¹ E a pomba voltou a ele à tarde; e eis, arrancada, uma folha de oliveira no seu bico; e conheceu Noé que as águas tinham minguado de sobre a terra.

> ¹² Então esperou ainda outros sete dias, e enviou fora a pomba; mas não tornou mais a ele.

> ¹³ E aconteceu que no ano seiscentos e um, no mês primeiro, no primeiro dia do mês, as águas se secaram de sobre a terra. Então Noé tirou a cobertura da arca, e olhou, e eis que a face da terra estava enxuta.

> ¹⁴ E no segundo mês, aos vinte e sete dias do mês, a terra estava seca.

> ¹⁵ Então falou Deus a Noé dizendo:

> ¹⁶ Sai da arca, tu com tua mulher, e teus filhos e as mulheres de teus filhos.

> ¹⁷ Todo o animal que está contigo, de toda a carne, de ave, e de gado, e de todo o réptil que se arrasta sobre a terra, traze fora contigo; e povoem abundantemente a terra e frutifiquem, e se multipliquem sobre a terra.

> ¹⁸ Então saiu Noé, e seus filhos, e sua mulher, e as mulheres de seus filhos com ele.

> ¹⁹ Todo o animal, todo o réptil, e toda a ave, e tudo o que se move sobre a terra, conforme as suas famílias, saiu para fora da arca. (Gênesis 8:1-19, ACF)

## Ato 8 — O recomeço
Depois do dilúvio, os personagens ajudam a estabilizar o novo começo: organização, escassez, temor e esperança. A lição é que Deus julga, mas também preserva e reordena a vida para um novo pacto. Os testes finais são de Sobrevivência, Ofício, Diplomacia e Vontade.

> ²⁰ E edificou Noé um altar ao Senhor; e tomou de todo o animal limpo e de toda a ave limpa, e ofereceu holocausto sobre o altar.

> ²¹ E o Senhor sentiu o suave cheiro, e o Senhor disse em seu coração: Não tornarei mais a amaldiçoar a terra por causa do homem; porque a imaginação do coração do homem é má desde a sua meninice, nem tornarei mais a ferir todo o vivente, como fiz.

> ²² Enquanto a terra durar, sementeira e sega, e frio e calor, e verão e inverno, e dia e noite, não cessarão. (Gênesis 8:20-22, ACF)

Efeitos escalados de fé
Para esta história, eu sugiro usar estes efeitos em toda cena importante:

1 símbolo de fé: uma ajuda pequena, coragem momentânea ou direção clara.

2 símbolos de fé: vantagem concreta; um obstáculo é contornado ou reduzido.

3 símbolos de fé: mudança forte na cena; a situação começa a pender a favor do grupo.

4 símbolos de fé: intervenção marcante; algo decisivo acontece sem quebrar o evento bíblico.

Esses efeitos mantêm a mesa sentindo que Deus age de modo real, mas sem permitir que os jogadores desfaçam o curso das Escrituras. Em cenas bíblicas inevitáveis, a fé não muda o fato central; ela muda a forma como o grupo atravessa a fidelidade, o medo e a responsabilidade.

Complicações escaladas
Sugestão de escada de complicações:

1 complicação: atraso, ruído, custo pequeno, desgaste emocional.

2 complicações: oposição social, perda parcial de recurso, ferimento leve, separação momentânea.

3 complicações: sabotagem, ruptura de grupo, ameaça séria ou risco de falha em cadeia.

4 complicações: colapso da cena; só a intervenção final do mestre ou um milagre raro evita desastre maior.

Na prática, complicação nunca deve impedir o evento bíblico central; ela deve aumentar a pressão para que os jogadores entendam o peso moral e espiritual do momento.

Como ensinar valores
A aventura ensina três camadas de aprendizado:

História: origem do pecado, juízo, preservação e recomeço da humanidade.

Valores bíblicos: obediência, arrependimento, perseverança, humildade e confiança em Deus.

Postura de mesa: o jogador descobre que agir bem nem sempre evita sofrimento, mas sempre molda o coração e preserva o propósito.

O melhor fechamento é mostrar que Noé não foi salvo por esperteza, e sim por obediência; e que os personagens aprenderam a cooperar com algo maior do que eles mesmos.
