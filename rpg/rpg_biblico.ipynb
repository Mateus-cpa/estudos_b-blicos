## Criação do Personagem

### Criação

Neste sistema, cada personagem representa uma pessoa inserida em um povo, chamada por Deus a agir com coragem, fidelidade e obediência em meio a desafios reais. O foco do jogo não está em poderes mágicos pessoais, mas em decisões difíceis, oração, confiança, sabedoria prática, liderança e perseverança.

Na criação do personagem, siga estas etapas:

- Escolha o povo ao qual o personagem pertence.

- Defina os atributos.

- Escolha um treinamento.

Aplique as perícias recebidas por atributos, povo e treinamento.

Registre as marcas de fé, compromisso e vocação do personagem.

O mestre pode permitir que certos personagens tenham funções espirituais especiais, como sacerdote, profeta ou levita, mas essas funções devem priorizar serviço, intercessão, conselho, adoração e fidelidade em vez de magia ofensiva.

### Progressão

Com o avanço da história, o personagem ganha mais dados de classe nas rolagens. Isso faz o personagem crescer em presença dramática, não apenas em força matemática.

- no início: 1 dado de classe em situações apropriadas;

- depois: 2 dados de classe;

- mais tarde: 3 dados de classe;

- raramente: 4 dados de classe em momentos decisivos.

Isso deixa o jogo cada vez mais épico sem quebrar a simplicidade.

## Estrutura dos dados de jogo


### ⬜ Dado comum
O dado comum é um dado narrativo de 6 faces, usado em todos os testes do jogo. Em vez de números, ele apresenta símbolos que indicam sucesso, anulação, complicação e outros efeitos especiais.


As faces do dado comum é:

| Tipo                | Face | Efeito                                                   |
| ------------------- | ---- | -------------------------------------------------------- |
| Sucesso          | ⭐    | 1 sucesso                                                |
| Anulação          | ❌    | 1 anulação  de sucesso                                              |
| Anulação          | ❌    | 1 anulação de sucesso                                               |
| Sucesso maior          | ☀️    | 2 sucessos                                               |
| Dificuldade          | +🟥  | Adiciona 1 dado de dificuldade                           |
| Fé          | +🟩  | Adiciona 1 dado de Fé                                    |



### 🟥 Dado de dificuldade
O dado de dificuldade representa pressão, risco, oposição, custo narrativo e desgaste. Ele surge como efeito de um teste com resultado de adicionar dadod e dificuldade.

São as faces do dado de dificuldade:

| Tipo                | Face | Efeito                                                   |
| ------------------- | ---- | -------------------------------------------------------- |
| Sem efeito |     | Nada acontece                                     |
| Dado comum | +⬜    | Adiciona dado comum à rolagem                                     |
| Complicação | ⚠️   | Aumenta o risco e efeito de Desvantagem   |
| Anulação | ❌    | Anula 1 sucesso                                               |
| Anulação + Complicação | ❌⚠️  | Efeitos de Anulação + Complicação                                 |
| Desvantagem | ⛔    | Recebe 1 **marcador de desvantagem**|

**Complicação** não anula sucesso, mas aumenta o custo da cena, prepara risco, atraso, exposição ou desgaste, e recebe um ⛔ **marcador de desvantagem** .

**Marcador de desvantagem** ⛔ deve ser utilizado no momento que o narrador indicar.

O narrador pode conceder dados de dificuldade quando houver ambiente hostil, oposição forte, escassez, doença, pressa, confusão, medo, terreno difícil ou perda de tempo ou rodada.

### 🟩 Dado de fé
O dado de Fé representa a intervenção favorável de Deus na narrativa. Ele não é controlado pelo personagem, mas recebido como resposta à fé, à obediência e ao momento dramático da história.

São as faces do dado de fé:

| Tipo                | Face | Efeito                                                   |
| ------------------- | ---- | -------------------------------------------------------- |
| Marcador de vantagem          | 🔵   | O jogador recebe 1 marcador de vantagem                  |
| Direção          | 👉🏼 | Jogador recebe uma pista, ajuda inesperada       |
| Proteção          | 🛡️  | Cancela 1 Complicação, dano ou Anulação                  |
| Milagre raro          | 🪽   | Milagre raro, efeito extraordinário do Treinamento       |
| Dado comum | +⬜    | Adiciona  dado comum à rolagem                                     |
| Sem efeito |     | Nada acontece                                     |


- 🔵 **Vantagem**: O jogador recebe um marcador de vantagem para utilizar à sua escolha durante a sessão.

- 👉🏼 **Direção**: O narrador revela a melhor ação, pista ou decisão, surge uma ajuda inesperada, oportunidade ou caminho aberto.

- 🛡️ **Proteção**: cancela uma ⚠️ Complicação, dano ou ❌ Anulação.

- 🪽 **Milagre raro**: quando este símbolo aparece, o narrador interpreta um efeito extraordinário relacionado ao Treinamento do personagem. O efeito deve ser maior quanto mais símbolos 🪽 forem obtidos na rolagem. Um símbolo concede um benefício pontual; dois, um resultado relevante; três, uma mudança decisiva; quatro ou mais, um milagre excepcional.

O dado de Fé pode aparecer quando o personagem manifesta confiança, obediência, oração, coragem ou fidelidade em situações decisivas.

## Estrutura do teste
Todo teste tem as seguntes etapas:
1. **Determinar tipo de teste**: se será normal, com vantagem ou desvantagem. O narrador determina se utilizará marcador de desvantagem e o jogador decide se utilizará vantagem.

2. **Rolagem do teste**:

3. **Contabilizar os resultados**

4. **Resolver os efeitos colaterais**


### 1} Determinando o tipo de teste

Todo teste começa com 3 dados comuns.

- Vantagem: +1 dado comum.

- Desvantagem: -1 dado comum.

- Treinamento: altos níveis de personagem dão acesso a dados de Fé desde a rolagem inicial.

Nenhum teste deve cair abaixo de 1 dado, salvo em situações extremamente severas definidas pelo narrador.

Marcadores de vantagem e desvantagem não se acumulam. Sempre vale apenas 1 marcador de cada tipo por vez.

Contudo, um efeito de vantagem pode anular um de desvantagem. E vice-versa.

#### **Função dos marcadores**
Existem distinções entre os tipos de marcadores:

#### Marcador de vantagem
- Pode ser guardado para uma rolagem futura.

- Só pode ser usado uma vez.

- Não acumula com outro marcador igual.

#### Marcador de desvantagem
- Deve ser aplicado quando o narrador julgar apropriado.

- Também não acumula.

- Pode ser usado para representar pressões persistentes da cena.

### 2} Rolando os dados

Quando houver rolagem de dados, algumas faces de dados pedem para adicionar novos dados. Sempre que ocorrer, role-os antes de encerrar a fase de rolagem de dados e agrupe-os por símbolos:

- Sucessos
- Anulações
- Complicações
- Proteções
- Marcadores de vantagem
- Marcadores de desvantagem
- Fé

### 3} Contabilizando o resultado líquido

Nesta etapa, calculamos os saldos líquidos em jogo

- Sucesso x Anulação;
- Complicação x Proteção;
- Marcadores de vantagem x de desvantagem.


### 4} Resolvendo os efeitos colaterais

Isso evita discussão na hora de interpretar uma rolagem com muitos símbolos.

### Efeitos de Milagre raros

O 🪽 Milagre raro terá efeitos mais forte conforme a quantidade de faces obtidas  cresce. Ele é escalado em quatro níveis:

- 1 símbolo 🪽: milagre menor, útil e pontual.

- 2 símbolos 🪽🪽: milagre relevante, muda a cena.

- 3 símbolos 🪽🪽🪽: milagre forte, altera o rumo do conflito.

- 4+ símbolos 🪽🪽🪽🪽: milagre excepcional, quase decisivo.

### Testes de resultados dos dados







```python
import random
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Configurações iniciais dos dados
comum = ["⭐", "❌", "", "☀️", "+🟥", "+🟩"]
dificuldade = ["+⬜", "⚠️", "", "❌⚠️", "⛔"]
fe = ["🔵", "👉🏼", "🛡️", "🪽", "+⬜", "", ""]

def rolar_dado(tipo_dado):
    return random.choice(tipo_dado)

def realizar_rolagem(tipo_teste):
    """
    Realiza uma rolagem completa de dados para um determinado tipo de teste.
    Retorna o saldo líquido da rolagem.
    """
    dados_a_rolar = [comum] * tipo_teste
    resultados_brutos = []

    while dados_a_rolar:
        dado_atual = dados_a_rolar.pop(0)
        resultado = rolar_dado(dado_atual)
        resultados_brutos.append(resultado)

        if resultado == "+🟥":
            dados_a_rolar.append(dificuldade)
        elif resultado == "+🟩":
            dados_a_rolar.append(fe)
        elif resultado == "+⬜":
            dados_a_rolar.append(comum)

    resultado_liquido = 0
    for i in resultados_brutos:
        if i == "⭐" or i == "🛡️":
            resultado_liquido += 1
        elif i in ["❌", "❌⚠️"]:
            resultado_liquido -= 1
        elif i == "☀️":
            resultado_liquido += 2

    return resultado_liquido
```

```python
# Configurações da simulação
quantidades = [100, 1000, 10000, 100000, 1000000]
tipos_teste = {2: "Desvantagem", 3: "Normal", 4: "Vantagem"}

resultados_gerais = []
taxas_sucesso = {nome: [] for nome in tipos_teste.values()}

print("Iniciando as simulações (isso pode levar de 10 a 20 segundos)...")

# Executando simulação
for q in quantidades:
    for tipo, nome in tipos_teste.items():
        sucessos = 0
        for _ in range(q):
            res = realizar_rolagem(tipo)
            # Armazena os dados para os boxplots
            resultados_gerais.append({'Quantidade': q, 'Teste': nome, 'Resultado Líquido': res})

            # Conta os sucessos para o gráfico de linhas
            if res > 0:
                sucessos += 1

        taxa = (sucessos / q) * 100
        taxas_sucesso[nome].append(taxa)

df_resultados = pd.DataFrame(resultados_gerais)
print("Simulação concluída com sucesso! Os dados estão prontos para os gráficos.")
```

```python
# Configurações visuais do Boxplot
plt.style.use('ggplot')
fig, axes = plt.subplots(1, 3, figsize=(18, 6), sharey=True)

tipos = ["Desvantagem", "Normal", "Vantagem"]

for i, nome in enumerate(tipos):
    df_filtrado = df_resultados[df_resultados['Teste'] == nome]

    # CORREÇÃO APLICADA AQUI: hue='Quantidade' e legend=False adicionados
    sns.boxplot(data=df_filtrado, x='Quantidade', y='Resultado Líquido', hue='Quantidade',
                ax=axes[i], palette="viridis", legend=False)

    axes[i].set_title(f'Distribuição - Teste {nome}')
    axes[i].set_xlabel('Quantidade de Rolagens')

    # Coloca o rótulo do eixo Y apenas no primeiro gráfico
    if i == 0:
        axes[i].set_ylabel('Resultado Líquido')
    else:
        axes[i].set_ylabel('')

    # Formata os labels do eixo X (ex: 1.000, 10.000)
    labels = [f'{q:,}'.replace(',', '.') for q in quantidades]
    axes[i].set_xticks(range(len(quantidades)))
    axes[i].set_xticklabels(labels, rotation=45)

plt.tight_layout()
plt.show()
```

```python
# Configurações visuais do Gráfico de Linhas
plt.figure(figsize=(10, 6))
cores = {'Desvantagem': '#e74c3c', 'Normal': '#3498db', 'Vantagem': '#2ecc71'}
tipos = ["Desvantagem", "Normal", "Vantagem"]

for nome in tipos:
    plt.plot(quantidades, taxas_sucesso[nome], marker='o', markersize=8,
             label=f'{nome} ({list(tipos_teste.keys())[tipos.index(nome)]} Dados)',
             color=cores[nome], linewidth=2.5)

# Aplica a escala logarítmica no eixo X
plt.xscale('log')

# Textos e rótulos
plt.title('Taxa de Sucesso (%) vs Quantidade de Rolagens', fontsize=16)
plt.xlabel('Quantidade de Rolagens (Escala Logarítmica)', fontsize=12)
plt.ylabel('Taxa de Sucesso (%)', fontsize=12)

# Ajusta a formatação dos números no eixo X para o padrão com pontos
plt.xticks(quantidades, [f'{q:,}'.replace(',', '.') for q in quantidades])

plt.legend(fontsize=12)
plt.grid(True, which="both", ls="--", alpha=0.7)

plt.show()
```

# Fases da História
1. Origem da humanidade
Este bloco cobre criação, queda, Caim e Abel, Noé e o Dilúvio, além da dispersão dos povos. Para o jogo, ele funciona como mito fundacional, explicando pecado, violência, pacto, julgamento e recomeço. É ótimo para introduções teológicas, mas ruim para campanha longa, porque há poucos personagens recorrentes e pouca geopolítica estável.

2. Patriarcas
Aqui entram Abraão, Isaque, Jacó e José, com a formação das famílias tribais e a ida ao Egito. Esse período é excelente para campanhas de viagem, alianças, conflitos familiares e promessas divinas, porque tudo ainda está em formação. É um bloco muito aplicável ao jogo, especialmente se você quiser jogar “origem de povo” e “vocação de família”.

3. Egito e opressão
Este bloco cobre o crescimento dos hebreus no Egito, a mudança política, a escravidão e a pressão imperial. Em termos de cenário, ele é forte para drama social, perseguição, trabalho forçado e resistência religiosa. Serve bem para aventuras de fuga, sobrevivência e conspiração, mas menos para exploração aberta de mapa.

4. Êxodo e deserto
Aqui entram Moisés, as pragas, a saída do Egito, o Sinai, a aliança e a caminhada no deserto. Esse é um dos melhores blocos para o seu sistema, porque combina escassez, pressão, fé, risco, liderança e milagres narrativos. Ele é muito aplicável tanto para campanha curta quanto para uma fase inteira do jogo.

5. Conquista e Juízes
Este bloco cobre Josué, a entrada em Canaã, a divisão das terras e o período dos juízes. É um período ideal para campanhas de fronteira, guerra irregular, ambiguidade moral e assentamento de povos. Aqui o jogo pode ficar mais tático e territorial, com aldeias, conflitos locais e alianças instáveis.

6. Monarquia unida
Aqui estão Saul, Davi e Salomão, com a centralização do poder, a conquista política, a formação da capital e o templo. Esse bloco é ótimo para intriga de corte, guerras de expansão, sucessão e construção de identidade nacional. É um período muito aplicável porque permite campanha épica, política e espiritual ao mesmo tempo.

7. Reino dividido
Depois de Salomão, o reino se parte em Israel ao norte e Judá ao sul, com dinastias, profetas e crises internas. Este é um bloco riquíssimo para campanha, porque mistura guerra civil, idolatria, conflitos de legitimidade e intervenção profética. Ele também é um dos melhores para o seu sistema, já que os treinamentos e povos podem influenciar fortemente o jogo social e espiritual.

8. Exílio e retorno
Este bloco cobre a queda de Samaria, a queda de Jerusalém, o cativeiro babilônico e o retorno sob domínio estrangeiro. Para a mesa, ele funciona muito bem como cenário de perda, memória, reconstrução e fidelidade em terra estranha. É um bloco excelente para histórias mais maduras, com foco em restauração, identidade e resistência cultural.

9. Período de transição persa
Se você quiser um nono bloco opcional, vale separar o retorno do exílio e a reorganização sob a Pérsia. Aqui entram reconstrução do templo, restauração comunitária e tensões entre pureza, lei e sobrevivência política. Esse período é ótimo para campanhas de reforma e reconstrução, e conversa bem com funções como sacerdote, escriba, profeta e conselheiro.

# Atributos

Todo personagem possui seis habilidades básicas: Força (FOR), Agilidade (AGI), Vigor (VGR), Intelecto (INT), Sabedoria (SAB) e Presença (PRS). Essas habilidades representam os talentos naturais do personagem e determinam sua eficiência em diferentes tipos de ação.

Ao criar um personagem, distribua os valores das habilidades conforme as regras da campanha. Depois disso, cada habilidade gera efeitos diretos sobre as perícias relacionadas.

| Atributo | Nome | Perícias |
| -- | -- | -- |
|FOR|Força | Luta, Trabalho |
|AGI|	Agilidade  | Arquearia, Furtividade, Acrobacia|
|VGR|	Vigor  | Resistência, Sobrevivência, Cura|
|INT|	Intelecto  | Teologia, História, Ofício|
|SAB| Sabedoria |	Percepção , Intuição, Vontade|
|PRS| Presença	| Diplomacia, Liderança, Performance|



## Modificadores de Atributo

Cada Atributo gera um modificador de atributo, usado para definir uma vantagem em perícia. Em vez de aplicar apenas números, o modificador indica em qual perícia o personagem é especialmente competente.

- FOR favorece ações de combate físico e esforço corporal.

- AGI favorece precisão, movimento e discrição.

- VGR favorece resistência, fôlego e sobrevivência física.

- INT favorece conhecimento, técnica e raciocínio.

- SAB favorece percepção, discernimento e firmeza interior.

- PRS favorece influência, liderança e expressão.



## Regra de Vantagem por Habilidade
Cada personagem recebe uma vantagem em perícia para cada ponto de modificador de habilidade positivo.
Essa vantagem deve ser aplicada em uma perícia coerente com a habilidade correspondente.

Exemplo:

- Um personagem com AGI alta pode receber vantagem em Furtividade ou Acrobacia.

- Um personagem com INT alta pode receber vantagem em Teologia, História ou Ofício.

- Um personagem com SAB alta pode receber vantagem em Percepção, Intuição ou Vontade.

Se o personagem tiver modificadores positivos em mais de uma habilidade, ele recebe vantagens adicionais, sempre respeitando a coerência temática.

# Treinamentos
Os treinamentos representam as vocações fundamentais do povo de Deus em meio à história. Cada personagem pode ser formado para conselhar, construir, guardar, lutar, servir, mensagear, pastorear, profetizar ou ministrar diante do Senhor. Essas funções não são apenas ocupações; são maneiras de viver a fé em ação, servindo ao próximo e dependendo da direção de Deus.

Eles mostram como o personagem foi formado, qual função ocupa e de que maneira ele serve o povo. Os treinamentos destacam vocações práticas, espirituais e comunitárias.


## Conselheiro
Os conselheiros aparecem em Israel como homens e mulheres de discernimento, chamados a sustentar o povo em tempos de incerteza. Sua força não está na espada, mas na palavra certa, no temor de Deus e na capacidade de ouvir antes de falar. Eles servem em disputas, decisões tribais e momentos em que a unidade vale mais do que a vitória imediata.

Treinamento voltado à sabedoria prática e liderança moral. Crucial em conflitos sociais, decisões difíceis e mediação.

### Referências bíblicas
- ¹⁹ Ouve agora minha voz, eu te aconselharei, e Deus será contigo. Sê tu pelo povo diante de Deus, e leva tu as causas a Deus;
²⁰ E declara-lhes os estatutos e as leis, e faze-lhes saber o caminho em que devem andar, e a obra que devem fazer. (Êxodo 18:19-20, ACF)
- ¹⁴ Não havendo sábios conselhos, o povo cai, mas na multidão de conselhos há segurança. (Provérbios 11:14, ACF)
1 Reis 12:6-8.
- ⁶ E teve o rei Roboão conselho com os anciãos que estiveram na presença de Salomão, seu pai, quando este ainda vivia, dizendo: Como aconselhais vós que se responda a este povo?
⁷ E eles lhe falaram, dizendo: Se hoje fores servo deste povo, e o servires, e respondendo-lhe, lhe falares boas palavras, todos os dias serão teus servos. (1 Reis 12:6-7, ACF)

### Função
Orientar, julgar, aconselhar, resolver conflitos e sustentar a coesão do grupo.
Voltado em sabedoria, decisão, unidade.

### Atributos
SAB, PRS, INT.

### Perícias
Intuição, Liderança, História.

## Artesão

Os artesãos são os que transformam matéria bruta em provisão, beleza e utilidade. No deserto e na terra prometida, eles sustentam o povo com mãos firmes, preparando tendas, instrumentos, armas, utensílios e obras santas. Seu chamado lembra Bezalel e Aoliabe, capacitados por Deus para servir com excelência.

São cruciais para fabricação, reparo, criação de objetos e estruturas.

⁴ Para elaborar projetos, e trabalhar em ouro, em prata, e em cobre,

#Perícias

As perícias representam o que o personagem aprendeu na prática. Enquanto as habilidades refletem potencial natural, as perícias mostram treinamento, experiência e especialização.

Quando um personagem tenta fazer algo arriscado, o mestre pode pedir um teste de perícia apropriado. Se o personagem tiver vantagem naquela perícia, ele rola com maior chance de sucesso.





## Vantagens em Perícias
Vantagem em perícia significa que o personagem é especialmente competente naquela área. No jogo, isso pode ser representado por:

rolar dois dados e usar o melhor resultado;

receber bônus narrativo;

reduzir dificuldade em situações compatíveis;

obter melhor posição em disputas sociais ou físicas.

Um personagem pode receber vantagem em perícia por três fontes:

- Atributo;

- Povo;

- Classe.

- Dádiva (talento)

Essas fontes podem se acumular se o sistema da campanha permitir. Caso duas ou mais fontes concedam vantagem na mesma perícia, o mestre pode optar por:

- manter apenas uma vantagem;

- transformar o excesso em benefício narrativo;

- ou conceder um efeito especial adicional, se desejar mais destaque ao personagem.



## Estrutura das perícias
No seu sistema simplificado, cada perícia representa um tipo de ação recorrente. A ideia é que o mestre peça um teste quando:

- houver risco real de falha;

- a ação for importante para a cena;

- o resultado mudar a história;

- houver oposição ativa de outra criatura ou personagem.

Em geral:

- **Sucesso**: o personagem consegue fazer o que tentou.

- **Sucesso alto**: consegue com vantagem narrativa, rapidez ou estilo.

- **Falha**: não consegue, ou consegue com custo.

- **Falha crítica**: pode gerar barulho, atraso, ferimento, perda de recurso ou suspeita.

### Força

- Arquearia: disparos à distância, mira precisa, uso de arco, funda ou armas de arremesso.

## Acrobacia (Agilidade)

Use para equilíbrio, saltos, rolamentos, escapar de agarrões e passar por lugares estreitos ou instáveis.

Também serve para:

- atravessar pedras, muros baixos ou telhados;

- escapar de queda;

- passar por terreno difícil com rapidez;

- fazer manobras corporais em fuga.

Exemplo: um mensageiro corre por uma encosta de pedras e precisa saltar entre duas saliências.

## Arquearia (Agilidade)

Use para atacar com armas de projétil e para ações que exijam mira à distância. Inclui arco, funda, dardo arremessado, lança em arremesso e, se existir no cenário, besta.

Também serve para:

- tiro preciso;

- acertar alvos pequenos;

- disparar montado;

- manter pressão à distância.

Exemplo: um midianita atira uma flecha em um animal antes que ele fuja.





## Cura (Vigor)
Use para primeiros socorros, tratamento de feridas, estabilizar feridos e lidar com enfermidades simples.

Também serve para:

- estancar sangramento;

- fazer curativos;

- tratar febre leve;

- cuidar de exaustão;

- aplicar ervas e remédios simples.

Exemplo: após uma escaramuça, um levita tenta salvar um ferido antes que ele piore.

## Diplomacia (Presença)
Use para negociar, convencer, apaziguar conflitos, pedir favores e falar com autoridade respeitosa.

Também serve para:

- tratar com povos estrangeiros;

- resolver disputas;

- obter passagem segura;

- fazer acordos;

- acalmar uma multidão.

Exemplo: um emissário pede aos edomitas permissão para atravessar uma estrada.





## Furtividade (Agilidade)

Use para se esconder, se mover sem ser notado, aproximar-se em silêncio e evitar atenção.

Também serve para:

- espionagem;

- entrar em campo inimigo;

- seguir alguém sem ser visto;

- esconder objetos ou pessoas;

- esperar em emboscada.

Exemplo: um espia cananeu tenta entrar no acampamento israelita sem ser visto.



## História (Intelecto)
Use para lembrar acontecimentos antigos, genealogias, guerras, alianças, reis, cidades e tradições dos povos.

Também serve para:

- reconhecer um povo pela origem;

- lembrar rotas e impérios;

- saber de batalhas passadas;

- entender rivalidades antigas;

- identificar ruínas e nomes antigos.

Exemplo: um ancião ouve o nome de uma cidade e lembra que ela já foi fortificada por um rei amorreu.



## Intuição (Sabedoria)
Use para perceber intenção, mentira, tensão social e emoção escondida.

Também serve para:

- saber se alguém está mentindo;

- sentir traição;

- entender clima de uma conversa;

- perceber medo ou culpa;

- ler postura e comportamento.

Exemplo: um juiz de Israel desconfia que um visitante não veio em paz.



## Luta (Força)
Use para combate corpo a corpo e agressão física direta. Isso inclui socos, quedas, golpes com espada, lança, cajado, clava e armas improvisadas de perto.

Também serve para:

- segurar alguém à força;

- abrir uma porta à força;

- empurrar, derrubar ou imobilizar;

Exemplo: um hebreu tenta derrubar um filisteu no chão durante uma emboscada.

- levantar ânimo;

## Ofício (Intelecto)

- reparar carroças, tendas e utensílios;

- trabalhar madeira, metal, couro e tecido;

- resistir a dor;

- continuar consciente após ferimento;

- ficar em guarda por horas.

Exemplo: a caravana precisa atravessar o deserto sem água suficiente.





## Sobrevivência (Vigor)

Use para viver e se orientar em ambientes hostis. É a perícia do deserto, da mata, da estrada e do acampamento.

Também serve para:

- encontrar água, comida e abrigo;

- seguir rastros;

- reconhecer clima e direção;

- montar acampamento seguro;

- perceber sinais de perigo natural.

- fazer plantações

Exemplo: um líder tribal procura um caminho seguro para uma rota de pastoreio.



## Teologia (Intelecto)
Use para entender a fé, os ritos, as leis sagradas, os mandamentos, os cultos e a interpretação religiosa dos acontecimentos.

Também serve para:

- reconhecer heresias;

- saber o sentido de festas e sacrifícios;

- interpretar sinais espirituais;

- conhecer sacerdócio e pureza ritual;

- lembrar histórias sagradas.

Exemplo: um sacerdote precisa decidir se um objeto pode entrar no tabernáculo.




## Trabalho (Força)
Use para esforço físico prolongado e tarefas pesadas. É a perícia de carregar, puxar, empurrar, erguer, cavar, cortar, construir e romper obstáculos.

Também serve para:

- mover pedra, madeira ou cargas;

- escavar trincheiras;

- abrir passagem;

- montar acampamento pesado;

- trabalho braçal repetitivo.

Exemplo: um grupo precisa levantar um abrigo no deserto antes do anoitecer.

## Vontade (Sabedoria)
Use para resistir a medo, intimidação, tentação, manipulação, loucura, opressão espiritual e pressão moral.

Também serve para:

- não quebrar sob ameaça;

- resistir a maldição ou encantamento;

- manter a fé;

- suportar visões aterradoras;

- não ceder a suborno moral.

Exemplo: um hebreu é pressionado a adorar um ídolo, mas tenta resistir.

## Como o mestre pode pedir testes
Uma forma simples de decidir:

- Luta: conflito físico direto.

- Trabalho: esforço braçal e construção.

- Arquearia: ataque ou ação à distância.

- Furtividade: esconder-se ou agir sem ser percebido.

- Acrobacia: movimento difícil e controle corporal.

- Resistência: suportar desgaste físico.

- Sobrevivência: viver no ambiente e se orientar.

- Cura: tratar feridos e doentes.

- Teologia: assuntos sagrados, ritos e fé.

- História: memória de povos, eventos e linhagens.

- Ofício: fabricar e consertar.

- Percepção: notar o que está acontecendo.

- Intuição: ler pessoas e intenções.

- Vontade: resistir psicologicamente ou espiritualmente.

- Diplomacia: conversar, negociar e persuadir.

- Liderança: comandar e inspirar.

- Performance: expressar, recitar, cantar.





## Regra prática para evitar confusão
Se a dúvida for entre duas perícias, use esta lógica:

É físico e direto? Luta.

É físico prolongado? Trabalho ou Resistência.

É sobreviver no ambiente? Sobrevivência.

É notar algo? Percepção.

É entender intenção? Intuição.

É falar com alguém? Diplomacia.

É comandar um grupo? Liderança.

É coisa sagrada? Teologia.

É lembrar fatos antigos? História.

É fabricar ou arrumar? Ofício.

Exemplo de cena
O grupo chega a uma aldeia cananeia.

O líder usa Diplomacia para pedir água.

O sentinela usa Percepção para notar guardas escondidos.

O batedor usa Furtividade para entrar na vila.

O artesão usa Ofício para reparar uma roda quebrada.

O profeta usa Teologia para interpretar um presságio.

O guerreiro usa Luta se a negociação falhar.

Posso transformar isso em uma tabela de referência rápida de uma página, ou já organizar as perícias em formato de ficha de personagem com exemplos de vantagem e desvantagem por tribo.

# (editar) Combate
O sistema de combate de rápido, narrativo e narração de resultados.

## Iniciativa

Se os jogadores estão cientes, todos jogadores rolam o teste de **Vontade**. Se forem surpreendidos, façam o teste de **Percepção**.

Quem tiver maior resultado líquido, jogará primeiro.

Se empatar, quem tiver mais símbolos de Direção ou Vantagem age primeiro.

Se ainda empatar, o narrador decide com base na ficção.

Uso narrativo
Percepção para notar o perigo primeiro.

Vontade para não hesitar sob ameaça.

Ataque padrão
No ataque, o jogador rola os 3 dados comuns.

Resultado do ataque
★ = 1 acerto.

☀ = 2 acertos.

✖ = cancela 1 acerto.

+✴️ = adiciona dado de dificuldade ao defensor ou à cena.

+🆙 = pode gerar dado de Fé, se a cena permitir.

Dano simples
Eu sugiro que o dano seja a diferença entre:

sucessos do atacante

menos defesas, anulações e proteções

Isso mantém o sistema fluido.

Defesa no combate
Aqui você pode usar duas opções, e eu acho que a melhor é permitir ambas conforme o tipo de defesa.

1. Defesa passiva
O defensor não rola nada.
O atacante só precisa superar a Defesa da posição do alvo.

Boa para:

manter ritmo rápido;

inimigos comuns;

cenas com muitos combatentes.

2. Defesa ativa
O defensor rola um teste antagônico de:

Acrobacia para esquivar;

Resistência para firmar postura, bloquear ou aguentar o impacto;

Luta se quiser aparar, conter ou lutar de perto.

Boa para:

chefes;

duelos;

momentos dramáticos.

Testes antagônicos
Eu faria o teste antagônico assim:

Regra base
Ambos rolam seus dados.

Cada lado soma seus sucessos líquidos.

Quem tiver mais sucessos vence.

A diferença define a intensidade do efeito.

Grau de vitória
1 sucesso de diferença: vitória leve.

2 sucessos de diferença: vitória clara.

3 sucessos de diferença: vitória decisiva.

4+ sucessos de diferença: vitória esmagadora.

Antagonismo no combate
Ataque vs Defesa
Exemplo:

Atacante rola 2 sucessos líquidos.

Defensor rola 1 sucesso líquido em Acrobacia.

Resultado final: 1 acerto.

Se o defensor tiver:

Proteção 🛡️, pode cancelar uma complicação, anulação ou parte do dano.

Direção 👉🏼, pode reposicionar-se ou escolher melhor a defesa.

Milagre raro 🪽, pode evitar uma consequência grave.

O papel da dificuldade no combate
Eu trataria o dado de dificuldade como o peso do combate.

Quando adicionar dificuldade
terreno ruim;

chuva, lama, areia, noite;

inimigos numerosos;

exaustão;

ferimentos prévios;

armas inadequadas;

escudo quebrado;

hesitação emocional;

perder 1 rodada.

Efeitos possíveis
reduzir sucessos;

gerar complicações;

criar desvantagem para a próxima ação;

impor atraso ou abertura tática.

Ferimentos
Em vez de pontos de vida complexos, eu usaria 3 ou 4 níveis de ferimento.

Exemplo simples
Sem ferimento

Ferido

Gravemente ferido

Caído

Incapacitado

Como aplicar
Cada ataque bem-sucedido pode:

causar 1 nível de ferimento;

ou piorar o ferimento existente, se a diferença for grande.

Uso dos símbolos
Complicação pode causar corte, perda de arma, tropeço ou sangramento.

Anulação cancela parte do golpe.

Proteção reduz ferimento.

Milagre raro pode impedir que um personagem caia.

Combate como cena
Uma boa forma de simplificar ainda mais é tratar o combate como cena em andamento, não como troca de números a cada golpe.

Rodada
Cada rodada um personagem escolhe uma ação:

atacar;

defender;

correr;

proteger aliado;

usar liderança;

buscar cobertura;

invocar fé / agir com propósito.

Efeito do sucesso
sucesso simples: faz a ação.

sucesso forte: faz com vantagem.

complicação: faz, mas paga preço.

anulação: falha.

milagre raro: efeito especial ligado ao treinamento.

Minha sugestão prática de combate
Eu faria assim:

Ataque
3 dados comuns.

Defesa
2 dados se a defesa for ativa.

1 dado se o personagem estiver desprevenido.

sem rolagem se for defesa passiva.

Resultado
sucessos líquidos do ataque - sucessos líquidos da defesa = resultado final.

Interpretação
0 ou menos: não acerta ou é totalmente bloqueado.

1: arranha, pressiona ou ferimento leve.

2: acerto sólido.

3: acerto forte.

4+: golpe decisivo.

Testes antagônicos fora do combate
Esse mesmo modelo serve para:

Diplomacia vs Intuição

Furtividade vs Percepção

Liderança vs Vontade

História vs História

Cura vs Resistência em situações de doença

Teologia vs Vontade para resistir a influência espiritual

Regra simples
quem tiver mais sucessos líquidos vence;

empate favorece quem tiver melhor posição narrativa;

em caso de empate perfeito, surge uma complicação ou escolha difícil.

Regras que eu evitaria
Eu evitaria:

bônus numéricos demais;

defesa automática muito alta;

muitos tipos de dano;

tabelas longas de armas;

subtração constante em toda rolagem.

Seu sistema ganha muito mais se o combate for:

rápido;

simbólico;

tenso;

com custo narrativo real.

Resumo da minha proposta
Combate
rolagem simbólica;

ataque e defesa antagônicos;

dificuldade gera custo;

ferimentos em estágios;

milagre raro como intervenção decisiva.

Testes antagônicos
cada lado rola seus dados;

compara sucessos líquidos;

diferença define intensidade;

complicações e fé alteram a cena.

Se quiser, eu posso transformar isso em uma regra formal de manual, já com texto pronto para o livro e uma seção específica de combate, defesa, ferimentos e testes antagônicos.

# Foco espiritual do jogo
Como você quer minimizar a temática mágica para os jogadores, o jogo pode tratar o sobrenatural de forma mais reverente e rara. Em vez de “magias”, os personagens têm:

atos de fé;

orações respondidas;

direção divina;

consolo e encorajamento;

proteção providencial;

sabedoria concedida por Deus em momentos críticos.

Isso mantém o cenário bíblico forte sem transformar os personagens em lançadores de feitiços. A sensação do jogo passa a ser: o personagem não controla o divino, ele depende do divino.

Se quiser, eu posso transformar isso agora em um texto ainda mais “editorial”, com linguagem de livro mesmo, já no formato final de manual, incluindo títulos, subtítulos e redação uniforme para colar direto no documento.
